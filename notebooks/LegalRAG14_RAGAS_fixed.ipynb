```python
def load_documents(path):

    loader = PyMuPDFLoader(path)

    docs = loader.load()

    return docs
```

Defines a function to ingest PDF files. PyMuPDFLoader is utilized because it is highly efficient at extracting clean text from complex legal layouts, preserving the structure better than standard PDF tools.

```python
def semantic_chunk(text: str, threshold: float = 0.78) -> list[str]:
    sentences = nltk.sent_tokenize(text)
    if len(sentences) <= 1:
        return sentences

    # BGE models benefit from the query prefix during encoding
    prefixed = [f"Represent this sentence for retrieval: {s}" for s in sentences]
    sent_embeddings = bge_sentence_model.encode(prefixed, normalize_embeddings=True)

    chunks, current = [], [sentences[0]]
    for i in range(1, len(sentences)):
        sim = float(np.dot(sent_embeddings[i], sent_embeddings[i-1]))  # already normalized
        if sim < threshold:
            chunks.append(" ".join(current))
            current = []
        current.append(sentences[i])
    chunks.append(" ".join(current))
    return chunks
```

### Semantic Chunking Logic

This function implements a "meaning-aware" splitting strategy. Instead of cutting text at a fixed character count, it:

Splits the document into sentences.

Computes the Cosine Similarity between embeddings of adjacent sentences.

Breaks the text into a new chunk only when the similarity falls below 0.75.

Why: This ensures that a legal concept or section remains in a single chunk, preventing the AI from losing context because a sentence was cut in half.

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=100
)

def adaptive_chunk(text):

    semantic_chunks = semantic_chunk(text)

    final_chunks = []

    for chunk in semantic_chunks:

        splits = splitter.split_text(chunk)

        final_chunks.extend(splits)

    return final_chunks
```

### Adaptive Recursive Chunking

This cell defines a multi-stage splitting strategy. It takes the "Semantic Chunks" and, if they are still too large (e.g., over 500 characters), it applies a RecursiveCharacterTextSplitter.

It includes a 100-character overlap to ensure that even when a chunk is split, the "flavor" of the previous paragraph is preserved for the model.

```python
def hierarchical_split(docs: list) -> tuple[list, list]:
    parent_docs, child_docs = [], []
    for doc in docs:
        parents = semantic_chunk(doc.page_content)
        for idx, parent_text in enumerate(parents):
            parent_meta = {
                **doc.metadata,
                "chunk_type": "parent",
                "parent_index": idx,
            }
            parent_docs.append(Document(page_content=parent_text, metadata=parent_meta))
            children = adaptive_chunk(parent_text)
            for child_text in children:
                child_docs.append(Document(
                    page_content=child_text,
                    metadata={
                        **doc.metadata,          # preserves page, file_path, source
                        "chunk_type": "child",
                        "parent_chunk": parent_text[:200],  # truncated reference
                    }
                ))
    return parent_docs, child_docs
```

Child Chunks: Smaller, highly specific snippets.

Logic: The system searches against the Child chunks for precision, but when it finds a match, it retrieves the Parent chunk to provide the LLM with the full surrounding context.

### Document Retriever Builder

This function automates the creation of the search infrastructure for an uploaded document:

Chroma DB: Creates a vector database in a local directory (./doc_chroma).

BM25 Index: Tokenizes the chunks to create a keyword-based search index.

This enables Hybrid Search (finding matches by both "meaning" and "specific keywords").

```python
# Load Existing Legal Vector DB
import re

# ── Vector store ─────────────────────────────────────────────────────────
DB_PATH    = os.getenv('CHROMA_PERSIST_DIR', '../data/vector_database')
COLLECTION = 'legal_knowledge_v2'

legal_vectorstore = Chroma(
    persist_directory=DB_PATH,
    embedding_function=embeddings,
    collection_name=COLLECTION,
)

docs = legal_vectorstore.get()

legal_corpus = docs["documents"]

def tokenize_legal(text: str) -> list[str]:
    """Lowercase, strip punctuation, remove stopwords for BM25."""
    tokens = re.findall(r'\b[a-zA-Z0-9]+\b', text.lower())
    # Preserve important legal abbreviations by skipping removal of short tokens
    return [t for t in tokens if len(t) > 1]

tokenized = [tokenize_legal(c) for c in legal_corpus]
legal_bm25 = BM25Okapi(tokenized)
```

### Knowledge Base Connection

This cell connects the system to a pre-existing "General Legal Database" stored on disk. It loads the vectors and the BM25 index for the entire Indian Legal corpus, allowing the AI to "know" the law even if the answer isn't in the user's uploaded file.

```python
# Reciprocal Rank Fusion
def rrf_fusion(vector_docs, bm25_docs, k=60):

    scores = {}

    for rank, doc in enumerate(vector_docs):
        key = doc.page_content
        scores[key] = scores.get(key, 0) + 1/(k+rank)

    for rank, doc in enumerate(bm25_docs):
        key = doc.page_content
        scores[key] = scores.get(key, 0) + 1/(k+rank)

    ranked = sorted(scores.items(), key=lambda x: x[1], reverse=True)

    return [Document(page_content=doc) for doc,_ in ranked]
```

### Reciprocal Rank Fusion (RRF)

Implements the RRF algorithm. Since we are performing two types of searches (Vector and BM25), we get two different lists of results. RRF uses a mathematical formula—$1 / (k + rank)$—to merge these lists into one perfectly sorted master list.

```python
def document_search(query, k=8):

    if document_vectorstore is None:
        return []

    vector_docs = document_vectorstore.similarity_search(query, k=k)

    scores = document_bm25.get_scores(query.split())

    idx = np.argsort(scores)[::-1][:k]

    bm25_docs = [Document(page_content=document_corpus[i]) for i in idx]

    return rrf_fusion(vector_docs, bm25_docs)
```

### Document Search Execution

The search_documents function:

Performs a similarity search on the uploaded file's vector store.

Performs a keyword search on the BM25 index.

Combines them using RRF to find the most relevant parts of the case file.

```python
def legal_search(query: str, k: int = 8) -> list:
    vector_docs = legal_vectorstore.similarity_search(query, k=k)
    scores = legal_bm25.get_scores(tokenize_legal(query))  # consistent tokenization
    idx = np.argsort(scores)[::-1][:k]
    bm25_docs = [Document(page_content=legal_corpus[i]) for i in idx]
    return rrf_fusion(vector_docs, bm25_docs)
```

### Query Rewrite Agent

```python
class LegalState(TypedDict):
    question: str
    uploaded_file: Optional[str]       # add this
    rewritten_query: Optional[str]
    strategy: Optional[str]
    legal_docs: Optional[List]
    document_docs: Optional[List]
    final_docs: Optional[List]
    answer: Optional[str]
    critique: Optional[str]
```

### Rewrite Node

The entry point of the graph. It calls the Rewrite Agent and updates the state with the optimized_query.

```python
import os
from typing import TypedDict, Optional

# 1. Resolve LLM dependencies
from langchain_groq import ChatGroq

# 2. Define the State
class AgentState(TypedDict):
    question: str
    uploaded_file: Optional[str]
    strategy: Optional[str]
    # Add any other keys your graph uses here (e.g., final_docs, answer)

# 3. Explicitly initialize the LLM
# FIX: Updated the model string to an active Groq Llama 3.1 endpoint
routing_llm = ChatGroq(
    model="llama-3.1-8b-instant", # Replaced the decommissioned 'llama3-8b-8192'
    temperature=0, 
    max_tokens=10
)

def strategist_agent(state: AgentState):
    print("---STRATEGIST AGENT---")
    question = state["question"]
    uploaded_file = state.get("uploaded_file")
    
    file_context = ""
    if uploaded_file:
        filename = os.path.basename(uploaded_file)
        clean_filename = filename.replace("_", " ").replace(".pdf", "").replace(".PDF", "")
        file_context = f"\nCRITICAL CONTEXT: An uploaded case document is available regarding '{clean_filename}'. If the user's question mentions this case, these parties, or asks for specifics not found in general law, you MUST route to DOCUMENT or BOTH."
    
    prompt = f"""You are an intelligent routing agent managing a legal RAG system.
    Analyze the following question and choose the appropriate retrieval strategy.
    
    Question: "{question}"
    {file_context}
    
    Options:
    - LEGAL: Use this ONLY for general Indian Law or Constitutional queries that do not relate to the uploaded document.
    - DOCUMENT: Use this if the query asks for specific details, plaintiffs, defendants, dates, or issues regarding the uploaded case document.
    - BOTH: Use this if the query requires combining general constitutional principles with the specific facts of the uploaded document.
    
    Return ONLY the exact word LEGAL, DOCUMENT, or BOTH. Do not include any other text, reasoning, or punctuation.
    """
    
    # 4. Use the correctly initialized LLM
    response = routing_llm.invoke(prompt)
    
    strategy = response.content.strip().upper()
    
    valid_strategies = ["LEGAL", "DOCUMENT", "BOTH"]
    if strategy not in valid_strategies:
        print(f"Warning: Unrecognized strategy '{strategy}'. Defaulting to safe route.")
        strategy = "BOTH" if uploaded_file else "LEGAL"
        
    print(f"Decision: Routed to {strategy}")
    return {"strategy": strategy}
```

### Strategist Node

The second node in the graph. It calls the Strategist Agent to decide the "Routing Path" (Legal, Document, or Both).

```python
def legal_agent(state):

    docs = legal_search(state["rewritten_query"])

    return {"legal_docs":docs}
```

### Legal Retrieval Node

A worker node that triggers if the strategy involves LEGAL. It calls search_legal_database and stores the findings in the state.

```python
def document_agent(state):

    docs = document_search(state["rewritten_query"])

    return {"document_docs":docs}
```

### Document Retrieval Node

A worker node that triggers if the strategy involves DOCUMENT. It calls search_documents and stores the findings in the state.

```python
def fusion_agent(state):

    docs = []

    if state.get("legal_docs"):
        docs.extend(state["legal_docs"])

    if state.get("document_docs"):
        docs.extend(state["document_docs"])

    docs = rerank(state["rewritten_query"], docs)

    return {"final_docs":docs}
```

### Fusion and Rerank Node

The "Aggregator" node. It takes all documents found by the previous nodes, removes duplicates, and applies the Cross-Encoder Reranker to find the absolute best context for the answer.

```python
def answer_agent(state: LegalState) -> dict:
    context = "\n\n".join([d.page_content for d in state["final_docs"]])
    question = state["question"]
    is_document_query = state.get("strategy") in ("DOCUMENT", "BOTH")

    citation_instruction = (
        "Cite the specific Article, Section, or case document clause that supports your answer."
        if not is_document_query else
        "Cite the specific paragraph, date, or party name from the uploaded case document."
    )

    prompt = f"""You are a precise legal assistant. Answer the question using ONLY the provided context.
If the context does not contain enough information, say so clearly — do not guess.
{citation_instruction}
Keep your answer concise and factual.

Context:
{context}

Question:
{question}

Answer:"""

    return {"answer": llm(prompt)}
```

### Answer Generation Node

The "Writer" node. It takes the final reranked context and the user's query. It generates a professional, grounded legal response using the Llama-3 model.

```python
def critic_agent(state):

    prompt=f"""
Evaluate the answer.

Question:
{state['question']}

Answer:
{state['answer']}

Is the answer correct and grounded?
Explain briefly.
"""

    critique = llm(prompt)

    return {"critique":critique}
```

### Critic Node

The "Fact-Checker" node. It reviews the generated answer against the retrieved context. If the answer contains "hallucinations" (info not found in the documents), it marks it for revision.

```python
def router(state):
    if state["strategy"] == "LEGAL":
        return "legal"
    elif state["strategy"] == "DOCUMENT":
        return "document"
    else:
        # Return the list of nodes directly from the router to trigger parallel execution
        return ["legal", "document"]
```

### Routing Logic (Conditional Edges)

This function defines the "Switchboard" of the graph. It looks at the strategy in the state and tells the graph whether to move to the Legal node, the Document node, or both.

```python
workflow = StateGraph(LegalState)

workflow.add_node("rewrite", rewrite_agent)
workflow.add_node("strategist", strategist_agent)
workflow.add_node("legal", legal_agent)
workflow.add_node("document", document_agent)
workflow.add_node("fusion", fusion_agent)
workflow.add_node("answer", answer_agent)
workflow.add_node("critic", critic_agent)

workflow.set_entry_point("rewrite")

workflow.add_edge("rewrite","strategist")

workflow.add_conditional_edges(
    "strategist",
    router,
    # Map the single string outputs. The list output will be handled natively by LangGraph.
    ["legal", "document"] 
)

workflow.add_edge("legal","fusion")
workflow.add_edge("document","fusion")
workflow.add_edge("fusion","answer")
workflow.add_edge("answer","critic")

graph = workflow.compile()
```
